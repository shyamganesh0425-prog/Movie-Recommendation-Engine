```python
import numpy as np
```

```python
movies = np.array(["Padayappa", "Adhi", "Goat", "Youth", "Karuppu", "Viswanath&Sons"])
ratings = np.array([
    [5, 1, 5, 0, 0, 0],
    [4, 2, 5, 5, 0, 0],
    [0, 5, 1, 4, 5, 0],
    [5, 0, 4, 5, 4, 0],
    [0, 5, 2, 4, 5, 0]
])

print(ratings)
```

```python
user_id = int(input("Enter User ID (1-5): "))
user_index = user_id - 1
user_ratings = ratings[user_index]
print("\nSelected User:", user_id)
print("Ratings:", user_ratings)
```

```python
similarity = np.zeros(len(ratings))

for i in range(len(ratings)):

    if i == user_index:
        similarity[i] = -1
        continue
    similarity[i] = np.dot(user_ratings, ratings[i]) / (
        np.linalg.norm(user_ratings) * np.linalg.norm(ratings[i])
    )
print("Similarity:", similarity)
```

```python
similar_users = np.argsort(similarity)[::-1]
print("Similar Users:", similar_users + 1)
```

```python
unwatched = np.where(user_ratings == 0)[0]
print("Unwatched Movies:")

for movie in unwatched:
    print(movies[movie])
```

```python
recommendation_scores = np.zeros(len(movies))

for user in similar_users:

    if user == user_index:
        continue

    for movie in unwatched:

        if ratings[user, movie] > 0:
            recommendation_scores[movie] += (
                similarity[user] * ratings[user, movie]
            )

print("Recommendation Scores:", recommendation_scores)
```

```python
sorted_movies = np.argsort(recommendation_scores)[::-1]

recommendations = []

for movie in sorted_movies:

    if movie in unwatched:
        recommendations.append(movie)

    if len(recommendations) == 3:
        break

print("\nTop 3 Recommended Movies:")

for i, movie in enumerate(recommendations, start=1):
    print(i, ".", movies[movie])
```
